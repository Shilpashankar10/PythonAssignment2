```python
student_mark={
    1: {'name': 'lando', 'mark': 75},
    2:{'name':'advik','mark':65},
    3:{'name':'shari','mark':85},
    4:{'name':'ridvik','mark':65},
    5:{'name':'chikku','mark':67}
    }
print('student dictionary:',student_mark)
```

```python
print("Lando's mark:",student_mark[1]['mark']) 
```

```python
student_mark.update({'name': 'Janani','mark':80})
print(student_mark)
```

```python
student_mark[1]['mark'] = 82
print(student_mark)
```

```python
student_mark_keys=student_mark.keys()
student_mark_values=student_mark.values()
student_mark_items=student_mark.items()
print('keys: ',student_mark_keys)
print('values: ',student_mark_values)
print('items: ',student_mark_items)
```

```python
my_set={'a','e','i','o','u','a','a','i'}
print('my_set:',my_set)
#set does not allow duplicate elements.So there are only 5 elements
```

```python
#my_set[4] = 's'-- we cannot modify the set elements using index because they are unordered
my_set.add('s')
print('my_set:',my_set)
```

```python
set1 = {1, 3, 5, 7, 9}  
set2 = {2, 3, 5, 8, 10} 
print('set1:',set1)
print('set2:',set2)
```

```python
intersectionset = set1 & set2
unionset = set1 | set2
print('intersection set:',intersectionset)
print('union set:',unionset)
```

```python
score = int(input('Enter your score(0 to 10):'))
if 0 <= score <= 10:
    if score > 7:
        print("Above Average: Great job! You're doing well. Keep up the good work and aim even higher!")
    elif 4 <= score <= 7:
        print("Average: Good effort! Keep practicing, there's room for improvement.")
    else:
        print("Below Average:Don't give up! Keep learning and practicing, there's plenty of room to grow.")
else:
    print("Invalid input! Please enter a score between 0 and 10.")
```
